# 01 - Similar Ticket Dataset Exploration

Explore `ml_training.historical_tickets` cluster fields and semantic groups for Task 7 retrieval.

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path.cwd().resolve().parents[3]
sys.path.insert(0, str(ROOT / "ml" / "benchmarks" / "similar_ticket_retrieval"))
from preprocessing import load_all_tickets, load_database_url, prepare_frames

all_tickets = load_all_tickets(load_database_url(ROOT))
queries, corpus = prepare_frames(all_tickets)
print(f"All tickets: {len(all_tickets)}")
print(f"Query split: {len(queries)} · Corpus: {len(corpus)}")
print(queries[["ticket_id", "similar_case_group_id", "incident_cluster", "semantic_group"]].head())
print("\nCluster sizes:\n", queries["similar_case_group_id"].value_counts().describe())
print("\nSemantic groups:\n", queries["semantic_group"].value_counts().head(10))
print("\nGround truth peers:\n", queries["ground_truth_ids"].map(len).describe())
```
